```python
#Goal 
#Automate the process of optimizing "w" and "b" using gradient Descet
```

```python
# Tools


# NumPy, a popular library for scientific computing
# Matplotlib, a popular library for plotting data
# plotting routines in the lab_utils.py file in the local directory
```

```python
import math, copy
import numpy as np
import matplotlib.pyplot as plt


```

```python
# Load data

# Simple synthetic data
x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([2.0, 4.0, 6.0, 8.0])
```

```python
# Define the model
def predict(x, w, b):
    return w * x + b
```

```python
#  Compute the Cost Function
# Mean Squared Error (MSE):
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost = (1/(2*m)) * np.sum((predict(x, w, b) - y)**2)
    return cost
```

```python
#Compute gradients
def compute_gradient(x, y, w, b):
    m = x.shape[0]
    dj_dw = (1/m) * np.dot((predict(x, w, b) - y), x)
    dj_db = (1/m) * np.sum(predict(x, w, b) - y)
    return dj_dw, dj_db
```

```python
#gradient descent algo
def gradient_descent(x, y, w_init, b_init, alpha, num_iters):
    w = w_init
    b = b_init
    cost_history = []

    for i in range(num_iters):
        dj_dw, dj_db = compute_gradient(x, y, w, b)

        w -= alpha * dj_dw
        b -= alpha * dj_db

        if i % 10 == 0:
            cost = compute_cost(x, y, w, b)
            cost_history.append(cost)
            print(f"Iteration {i}: Cost {cost:.4f}, w {w:.4f}, b {b:.4f}")

    return w, b, cost_history
```

```python
w_init = 0
b_init = 0
alpha = 0.1
iterations = 100

w_final, b_final, cost_history = gradient_descent(x, y, w_init, b_init, alpha, iterations)

print(f"\nFinal parameters: w = {w_final}, b = {b_final}")
```

```python
plt.scatter(x, y, color='red', label='Data')
plt.plot(x, predict(x, w_final, b_final), label='Regression Line')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression with Gradient Descent')
plt.grid(True)
plt.show()
```
